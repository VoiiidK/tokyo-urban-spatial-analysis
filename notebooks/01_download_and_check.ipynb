```python
import osmnx as ox
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt

print("All packages imported successfully.")
```

```python
place = "Asagaya Station, Suginami, Tokyo, Japan"

center_point = ox.geocode(place)

G = ox.graph_from_point(
    center_point,
    dist=1000,
    dist_type="network",
    network_type="walk"
)

fig, ax = ox.plot_graph(
    G,
    node_size=0,
    edge_linewidth=0.7,
    edge_color="black",
    bgcolor="white"
)
```
